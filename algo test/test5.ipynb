# Now let's see what is the profit


```python
import pandas as pd
profit = pd.read_csv('/Users/dhairya/cs projects/DS303 final project/dataset/profits.csv')
```

```python
profit['Total'] = 0
```

```python
profit['Total'] = profit.select_dtypes(include=['float64', 'int64']).sum(axis=1)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,6), dpi=300)

sns.set_palette("deep")
plt.plot(profit['Date'], profit['Total']/100000, linewidth=2.5, marker='o', 
         markersize=6, markerfacecolor='white', markeredgewidth=2)

plt.title('Total Profit Over Time', fontsize=18, pad=20, fontweight='bold')
plt.xlabel('Date', fontsize=14, labelpad=12, fontweight='bold')
plt.ylabel('Total Profit (in Lakhs)', fontsize=14, labelpad=12, fontweight='bold')

plt.grid(True, linestyle='--', alpha=0.3)
plt.xticks(rotation=45, ha='right', fontsize=11)
plt.yticks(fontsize=11)

plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['left'].set_linewidth(1.5)
plt.gca().spines['bottom'].set_linewidth(1.5)
plt.gca().xaxis.set_major_locator(plt.MaxNLocator(8))

plt.fill_between(profit['Date'], profit['Total']/100000, alpha=0.15, 
                 color=plt.gca().lines[-1].get_color())

plt.margins(x=0.02)
plt.tight_layout()
plt.show()
```
