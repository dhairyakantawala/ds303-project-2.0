# Mathematical Steps for Clustering Analysis

## 1. Normalized Prices
For each stock i at time t, we normalize the price series using:

P̃ᵢ(t) = (Pᵢ(t) - μᵢ) / σᵢ

Where:
- P̃ᵢ(t) is the normalized price
- Pᵢ(t) is the original price
- μᵢ is the mean price for stock i
- σᵢ is the standard deviation
- μᵢ = (1/T)∑ₜPᵢ(t) where T is total time periods

## 2. Cluster Assignment
For each cluster c, we define the set of stocks in that cluster:

Sₖ = {i : c(i) = k}

Where:
- Sₖ is the set of stocks in cluster k
- c(i) is the cluster assignment for stock i

## 3. Centroid Calculation
For each cluster c, we calculate the centroid series:

P̄ₖ(t) = (1/|Sₖ|)∑ᵢ∈Sₖ P̃ᵢ(t)

Where:
- P̄ₖ(t) is the centroid series for cluster k
- |Sₖ| is the number of stocks in cluster k
- P̃ᵢ(t) are the normalized prices of stocks in cluster k

## Signal Generation

### Cluster Spread (σc):
For each cluster c at time t:

σc(t) = √(1/|Sc| ∑(Pi(t) - P̄c(t))²)

where:
- |Sc| = number of stocks in cluster c
- Pi(t) = normalized price of stock i at time t  
- P̄c(t) = centroid price of cluster c at time t

### Z-Score (zi):
zi(t) = (Pi(t) - P̄c(t))/σc(t)

### Trading Signals:
- zi(t) > 3  → Short stock i
- zi(t) < -3 → Buy stock i 
- -3 ≤ zi(t) ≤ 3 → Hold


```python
z_scores_dict = {}

for k in range(5):
    cluster_stocks = cluster_cols[k]
    centroid = combined_df[f'cluster_{k}_centroid']
    spread = combined_df[f'cluster_{k}_spread']
    
    cluster_z_scores = {}
    for stock in cluster_stocks:
        stock_price = normalized_price_data[stock]
        z_score = (stock_price - centroid) / spread
        cluster_z_scores[stock] = z_score
        
    z_scores_dict.update(cluster_z_scores)

z_scores = pd.DataFrame(z_scores_dict, index=normalized_price_data.index)
z_scores
```
