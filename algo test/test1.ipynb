# Log Return Matrix Analysis

We will analyze the stock data by constructing a log return matrix with the following parameters:

- N (number of stocks in taken)
- T (number of time steps in our data)
- $P_i(t)$: price of stock i at time t

For each stock i, we calculate the log return at time t as:

$R_i(t) = \log P_i(t) - \log P_i(t-1)$

This gives us a log return matrix $R ∈ ℝ^{(T-1)×N}$:

$R = [R_i(t)]_{t=2}^T$

The log return matrix helps us analyze the relative price changes and correlations between different stocks over time.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

initial_data = pd.read_csv('/Users/dhairya/cs projects/DS303 final project/dataset/stocks_df.csv')
```

```python
stock_names = initial_data['Stock'].unique()
```

```python
log_returns_list = []
total_stocks = len(stock_names)
processed = 0

for stock in stock_names:
    stock_data = initial_data[initial_data['Stock'] == stock]
    log_returns = pd.DataFrame({'Date': stock_data['Date'], f'{stock}_log_return': np.log(stock_data['Close']).diff()})
    log_returns = log_returns.dropna()
    log_returns = log_returns.set_index('Date')
    log_returns = log_returns.loc['2020-01-01':'2022-01-01']
    if not log_returns.empty:
        log_returns_list.append(log_returns)
    print(log_returns)
    processed += 1
    remaining = total_stocks - processed
    print(f"Processed {processed}/{total_stocks} stocks. {remaining} stocks remaining.")
```

```python
log_returns_list_new = [df for df in log_returns_list if len(df) == 499]
```

```python
log_returns_matrix = pd.concat(log_returns_list_new, axis=1)
log_returns_matrix = log_returns_matrix.dropna(how='any')
log_returns_matrix.to_csv('/Users/dhairya/cs projects/DS303 final project/dataset/log_returns_matrix.csv')
```
