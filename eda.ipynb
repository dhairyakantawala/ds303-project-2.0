```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import matplotlib.pyplot as plt
import mplfinance as mpf
import plotly.graph_objects as go
```

```python
nifty_500 = pd.read_csv("C:/Users/advait/Python_VS_Code/DS303/stocks_df.csv/nifty500_reversed.csv")
nifty_500.head()
```

```python
nifty_500.isnull().sum()
```

```python
nifty_500.fillna(method = 'ffill', inplace = True)
```

```python
plt.figure(figsize=(10,5))

# plot High
plt.plot(nifty_500['Date'], nifty_500['High'],
         label='High Price',
         color='blue',
         alpha=0.8)

# plot Low
plt.plot(nifty_500['Date'], nifty_500['Low'],
         label='Low Price',
         color='orange',
         alpha=0.8)

# formatting
plt.xlabel('Time')
plt.ylabel('Value')
plt.title('Nifty 500: High vs Low Price')
plt.xticks(np.arange(0, 569, 100))    # your custom ticks
plt.legend()                          # show labels
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))

# plot High
plt.plot(nifty_500['Date'], nifty_500['Close'],
         label='Close Price',
         color='purple',
         alpha=0.8)

# plot Low
plt.plot(nifty_500['Date'], nifty_500['Open'],
         label='Open Price',
         color='pink',
         alpha=0.8)

# formatting
plt.xlabel('Time')
plt.ylabel('Value')
plt.title('Nifty 500: Close vs Open Price')
plt.xticks(np.arange(0, 569, 100))    # your custom ticks
plt.legend()                          # show labels
plt.tight_layout()
plt.show()
```

```python
plt.bar(nifty_500['Date'], nifty_500['Volume'],
         color='blue',
         alpha=0.8)
plt.xlabel('Time')
plt.xticks(np.arange(0, 569, 100))    # your custom ticks
#plt.yticks(np.arange(0, 569, 50))
plt.ylabel('Volume')
plt.title('Volume trend')
plt.tight_layout()
plt.show()
```
